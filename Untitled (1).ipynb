```python
# 1. Environment Setup & Working Directory Configuration
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
import statsmodels.formula.api as smf
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Suppress warnings
warnings.filterwarnings("ignore")

# Set working directory safely
directory_path = r"D:\Hon's 3rd year files"

try:
    if os.path.exists(directory_path):
        os.chdir(directory_path)
        print("Working directory successfully changed to:", os.getcwd())
    else:
        print(f"Directory '{directory_path}' not found. Using current working directory:", os.getcwd())
except Exception as e:
    print(f"Note on directory: {e}")
```

```python
# 2. Data Loading & Inspection
# Automatically find the first Excel or CSV file in the directory if file name is unknown
data_files = [f for f in os.listdir('.') if f.endswith(('.xlsx', '.xls', '.csv'))]

if data_files:
    file_name = data_files[0]
    print(f"Automatically detected data file: {file_name}")
else:
    file_name = "disease_dataset.xlsx" # Fallback file name

try:
    if file_name.endswith('.csv'):
        df = pd.read_csv(file_name)
    else:
        df = pd.read_excel(file_name)
    print("Dataset loaded successfully!")
except Exception as e:
    print(f"Error loading file: {e}")

# Basic inspections (Make sure df exists)
if 'df' in locals():
    print("\n--- Data Head ---")
    display(df.head())

    print("\n--- Data Info ---")
    df.info()

    print("\n--- Missing Values ---")
    print(df.isnull().sum())
else:
    print("DataFrame 'df' is not defined. Please check your file path and name.")
```

```python
# 3. Data Preprocessing & Feature Engineering
if 'df' in locals():
    # Fix potential column name typos (e.g., Distirct -> District)
    if "Distirct" in df.columns:
        df.rename(columns={"Distirct": "District"}, inplace=True)

    # Create target count variable by summing disease case columns (e.g., y1 to y6) if Y18 is missing
    if 'Y18' not in df.columns:
        case_cols = [col for col in df.columns if col.lower().startswith('y') and col[1:].isdigit() and int(col[1:]) <= 6]
        if case_cols:
            df['Y18'] = df[case_cols].sum(axis=1)
            print(f"Created 'Y18' by summing columns: {case_cols}")
        else:
            # Fallback if y1-y6 columns aren't found, pick a numerical column as target
            num_cols = df.select_dtypes(include=[np.number]).columns
            if len(num_cols) > 0:
                df['Y18'] = df[num_cols[-1]]
                print(f"Using '{df.columns[-1]}' as target column 'Y18'.")

    print("\n--- Updated Data Head ---")
    display(df.head())
else:
    print("Please run Cell 2 first to load the dataframe 'df'.")
```

```python
# 4. Full-Dataset GLM Modeling (Poisson vs. Negative Binomial)
if 'df' in locals() and 'Y18' in df.columns:
    # Dynamically check available columns for formula
    indep_vars = [col for col in ['x1', 'x2', 'x3', 'x4', 'x5', 'x6', 'x7', 'Year'] if col in df.columns]
    district_part = " + C(District)" if 'District' in df.columns else ""
    
    formula_str = f"Y18 ~ {' + '.join(indep_vars)}{district_part}"
    print(f"Using formula: {formula_str}")

    # Fit Poisson GLM
    poisson_model = smf.glm(formula=formula_str, data=df, family=sm.families.Poisson()).fit()
    print("\n--- Poisson GLM Summary ---")
    print(poisson_model.summary())

    # Overdispersion Test
    pearson_chi2 = np.sum(poisson_model.resid_pearson**2)
    df_resid = poisson_model.df_resid
    dispersion_stat = pearson_chi2 / df_resid

    print(f"\nPearson Chi-Square: {pearson_chi2:.2f}")
    print(f"Residual Degrees of Freedom: {df_resid}")
    print(f"Dispersion Statistic: {dispersion_stat:.2f}")

    if dispersion_stat > 1.5:
        print("⚠️ Overdispersion detected! Negative Binomial model is recommended over Poisson.")
    else:
        print("✓ No severe overdispersion detected.")

    # Fit Negative Binomial GLM
    nb_model = smf.glm(formula=formula_str, data=df, family=sm.families.NegativeBinomial()).fit()
    print("\n--- Negative Binomial GLM Summary ---")
    print(nb_model.summary())
else:
    print("Dataframe or target variable 'Y18' not available.")
```

```python
# 5. Model Comparison (AIC, Deviance, BIC)
if 'poisson_model' in locals() and 'nb_model' in locals():
    comparison_df = pd.DataFrame({
        'Model': ['Poisson GLM', 'Negative Binomial GLM'],
        'AIC': [poisson_model.aic, nb_model.aic],
        'Deviance (G^2)': [poisson_model.deviance, nb_model.deviance],
        'Log-Likelihood': [poisson_model.llf, nb_model.llf],
        'BIC': [
            -2 * poisson_model.llf + len(poisson_model.params) * np.log(poisson_model.nobs),
            -2 * nb_model.llf + len(nb_model.params) * np.log(nb_model.nobs)
        ]
    })

    print("\n--- Model Comparison Table ---")
    display(comparison_df)
else:
    print("Models need to be fitted first in Cell 4.")
```

```python
# 6. Train-Test Split & Out-of-Sample Forecasting
if 'df' in locals() and 'Year' in df.columns:
    target_year = df['Year'].max() # Automatically select the latest year as target year
    print(f"Target forecasting year set to: {target_year}")

    train = df[df['Year'] < target_year]
    test = df[df['Year'] == target_year]

    print(f"Train set observations: {len(train)}")
    print(f"Test set observations: {len(test)}")

    if len(test) > 0 and len(train) > 0:
        # Fit Negative Binomial model on training data
        forecast_model = smf.glm(formula=formula_str, data=train, family=sm.families.NegativeBinomial()).fit()

        # Predict on test data
        test = test.copy()
        test["Predicted_Cases"] = forecast_model.predict(test)

        preview_cols = ['Year'] + ([ 'District'] if 'District' in test.columns else []) + ['Y18', 'Predicted_Cases']
        print("\n--- Prediction Preview ---")
        display(test[preview_cols].head(20))
    else:
        print("Not enough data to split into train and test sets for the given year condition.")
else:
    print("DataFrame or 'Year' column missing.")
```

```python
# 7. Forecast Evaluation & Visualization
if 'test' in locals() and 'Predicted_Cases' in test.columns and len(test) > 0:
    actual = test['Y18']
    predicted = test['Predicted_Cases']

    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    r2 = r2_score(actual, predicted)

    print(f"Evaluation Metrics:")
    print(f" - MAE  : {mae:.4f}")
    print(f" - RMSE : {rmse:.4f}")
    print(f" - R²   : {r2:.4f}")

    # Scatter Plot: Actual vs Predicted
    plt.figure(figsize=(9, 6))
    sns.scatterplot(x=actual, y=predicted, alpha=0.7, color='teal', s=80)
    plt.plot([actual.min(), actual.max()], [actual.min(), actual.max()], 'r--', lw=2, label='Ideal Fit (y = x)')
    plt.xlabel("Actual Cases", fontsize=12)
    plt.ylabel("Predicted Cases", fontsize=12)
    plt.title("Actual vs Predicted Disease Cases", fontsize=14, fontweight='bold')
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.show()
else:
    print("Test predictions are not available yet.")
```
